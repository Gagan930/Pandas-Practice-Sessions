## Apply Method

```python
# Using apply we can apply any  user defined function on the series or dataframe

def marks(x):
    return x / 2
```

```python
df["Marks"].apply(marks)
```

```python
# lets make a new column for these half marks

df["Half marks"] = df["Marks"].apply(marks)

df
```

```python
# lambda function with apply

df["Marks"].apply(lambda x : x/2)
```

```python
# to get length of every name in the "Name" column

df["Name"].apply(len)
```

## Map function

```python
# we will perform encoding , will convert male to 1 and female to 0

df["Gender"].map({"Male" : 1, "Female" : 0})
```

```python
df["Encoded Gender"] = df["Gender"].map({"Male" : 1, "Female" : 0})
```

```python
df
```
